## 折线图的绘制 查看趋势


```python
# 绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

#创建图表 设置大小
plt.figure(figsize=(10,5))
month=['1月','2月','3月','4月']
sales = [100,150,80,130]

# 绘制折线图
plt.plot(month,sales,
         label='产品A',
         color='blue',
         linewidth=2,
         linestyle='--',
         marker='o', )

# 添加标题
plt.title('2025年销售趋势',color='green',fontsize=20)
# 添加坐标轴的标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('销售额（万元）',fontsize=10)
# 添加图例
plt.legend(loc='upper left',fontsize=10)
# 添加网格线
plt.grid(True,alpha=0.3,color='red',linestyle='--') #alpha是透明度 linsestyle线格式
# plt.grid(axis='y')
# 设置刻度字体大小
plt.xticks(rotation=0,fontsize=10) # rotation设置字体倾斜度
plt.yticks(fontsize=10)
# 设置y的范围
plt.ylim(0,160)
# 在每个数据点上显示数值
for x,y in zip(month,sales):
    print(x,y)
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)
# 显示图表
plt.show()
```

## 条形图和柱状图的绘制

```python
# 绘制柱状图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

#创建图表 设置大小
plt.figure(figsize=(10,5))
subjects=['语文','数学','英语','科学']
scores = [90,70,80,88]

# 绘制柱状图
plt.bar(subjects,scores,
label='小红',
color='red',
    width=0.6)

# 添加标题
plt.title('成绩',color='green',fontsize=20)
# # 添加坐标轴的标签
plt.xlabel('科目',fontsize=10)
plt.ylabel('分数',fontsize=10)
# 添加图例
plt.legend(loc='upper left',fontsize=10)
# # 添加网格线
plt.grid(axis='y',alpha=0.3,color='red',linestyle='--') #alpha是透明度 linsestyle线格式
# # plt.grid(axis='y')
# # 设置刻度字体大小
plt.xticks(rotation=0,fontsize=10) # rotation设置字体倾斜度
plt.yticks(fontsize=10)
# 设置y的范围
plt.ylim(0,100)
# # 在每个数据点上显示数值
for x,y in zip(subjects,scores):
    print(x,y)
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)

# 自动化优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制条形图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

# #创建图表 设置大小
plt.figure(figsize=(10,5))
subjects=['语文','数学','英语','科学']
scores = [90,70,80,88]
# # 绘制条形图
plt.barh(subjects,scores,
         color='red',)

# 添加标题
plt.title('成绩',color='green',fontsize=20)
# # # 添加坐标轴的标签
plt.xlabel('分数',fontsize=10)
plt.ylabel('科目',fontsize=10)
# 自动化优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

## 饼图和环形图的绘制

```python
# 绘制饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

#创建图表 设置大小
plt.figure(figsize=(10,5))
# 数据
things = ['学习','娱乐','运动','睡觉','其他']
times=[6,3,1,8,5]
color=['',''] #配色方案
# 绘制饼图
plt.pie(times,labels=things,
        autopct='%1.1f%%', # 显示百分比
        startangle=90) # autopct 显示%比 start....调整角度

# # 添加标题
plt.title('时间安排',color='green',fontsize=20)

# 自动化优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
## 环形图
# 绘制环形图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

#创建图表 设置大小
plt.figure(figsize=(10,5))
# 数据
things = ['学习','娱乐','运动','睡觉','其他']
times=[6,3,1,8,5]
# color=['',''] #配色方案
# 绘制环形图
plt.pie(times,labels=things,
        autopct='%1.1f%%', # 显示百分比
        startangle=90,#start....调整角度
        # color=color,
        wedgeprops={'width':0.3}, # wed...设置圆环
        pctdistance=0.8) #   #百分比离圆心的距离

# # 添加标题
plt.title('时间安排',color='green',fontsize=20)
plt.text(0,0,'总计：\n100%',ha='center',va='center',fontsize=20)
# 自动化优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制爆炸式的饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

#创建图表 设置大小
plt.figure(figsize=(10,5))
# 数据
things = ['学习','娱乐','运动','睡觉','其他']
times=[6,3,1,8,5]
# color=['',''] #配色方案
explode=[0.1,0,0,0,0] # 设置突出块的位置 离圆心距离
# 绘制饼图
plt.pie(times,labels=things,
        autopct='%1.1f%%', # 显示百分比
        startangle=0 # autopct 显示%比 start....调整角度
        ,explode=explode # 设置突出块
        ,shadow=True,) # 设置阴影
# # 添加标题
plt.title('时间安排',color='green',fontsize=20)

# 自动化优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

## 散点图

```python
# 绘制散点图
import random
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='simHei'#设置中文字体

#创建图表 设置大小
plt.figure(figsize=(10,8))
#数据
x=[]
y=[]
for i in range(1000):
    tmp = random.uniform(0,10)
    x.append(tmp)
    tmp2=2*tmp+random.randint(0,10)
    y.append(tmp2)
# 绘制散点图
plt.scatter(x,y,
color='red',
        alpha=0.5, # 透明度
        s=20, #圆点大小
        label='数据'
            )

# # 添加标题
plt.title('散点图',color='green',fontsize=20)
# # # 添加坐标轴的标签
plt.xlabel('x',fontsize=10)
plt.ylabel('y',fontsize=10)
# # 添加图例
plt.legend(loc='upper left',fontsize=10)
# # # 添加网格线
plt.grid(alpha=0.3,color='red',linestyle='--') #alpha是透明度 linsestyle线格式
# # # plt.grid(axis='y')
# # # 设置刻度字体大小
plt.xticks(rotation=0,fontsize=10) # rotation设置字体倾斜度
plt.yticks(fontsize=10)
# # 设置y的范围
plt.ylim(0,40)
# # # 在每个数据点上显示数值
# for x,y in zip(subjects,scores):
#     print(x,y)
#     plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)

# 画一个回归线
plt.plot([0,10],[0,20],color='orange')
# 自动化优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

## 箱线图

```python
# 箱线图
import matplotlib.pyplot as plt

# 模拟3门课的成绩
data = {
    '语文':[82,85,88,70,90,76,84,83,95],
    '数学':[75,80,79,93,88,83,87,89,92],
    '英语':[70,72,68,65,78,80,85,90,95]
}
plt.figure(figsize=(8,6))
plt.boxplot(data.values(),tick_labels=data.keys())

plt.title('成绩分布')
plt.ylabel('分数')
plt.grid(True,axis='y',linestyle='--',alpha=0.3)

plt.show()
```

```python
"""
折线图：趋势随时间变
条形图/柱状图：类别之间对比
饼图：整体组成比例
散点图：两变量相关性
箱线图：数据分布，异常
"""
```

## 多个图的绘制

```python
import matplotlib.pyplot as plt
# 数据
month=['1','2','3','4']
sales = [100,150,80,130]
# 创建画布 生成图
f1 = plt.subplot(2,2,1) #生成子图 行 列 索引
f1.plot(month,sales)
f2 = plt.subplot(2,2,2) # 最后面数字需要改变
f2.bar(month,sales)
f3 = plt.subplot(2,2,3)
f3.barh(month,sales)
f4 = plt.subplot(2,2,4)
f4.scatter(month,sales)
```

```python
# 分析案例：温度分析
# 1导入库
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='microsoft yahei'
# 2.导入数据
df =pd.read_csv('../datset/weather.csv')
df.head()

# 绘制一个气温的趋势变化图
df['date']=pd.to_datetime(df['date']) # 处理日期
df= df[df['date'].dt.year==2015]
plt.figure(figsize=(20,10))

plt.plot(df['date'],df['temp_max'],label='最高气温',color='red')
plt.plot(df['date'],df['temp_min'],label='最低气温',color='blue')
plt.title('2025年气温趋势',color='red',fontsize=20)
plt.xlabel('日期')
plt.ylabel('气温')
plt.legend(loc='upper left',fontsize=20) # 图例
plt.show()
# df.info()
# 绘制一个降水量的直方图
plt.hist(df['precipitation'],bins=5)
```
