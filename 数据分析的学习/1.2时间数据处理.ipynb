```python
# 时间数据处理
import pandas as pd
d = pd.Timestamp('2015-05-02 10:22')
print(d)
print(type(d))
print('年：',d.year)
print('月：',d.month)
print(d.day)
print(d.hour)
print('季度',d.quarter)
print('是否是月底',d.is_month_end)
## 方法
print('星期几',d.day_name())
print('转换为天',d.to_period('D')) # 转换为天 计算天
print('转换为季度',d.to_period('Q')) # 转换为季度 可以计算季度的收入什么的
print('转换为年度',d.to_period('Y')) # 转换为年度
```

```python
# 字符串转换为日期类型
a = pd.to_datetime('2015-05-02 10:22')
a = pd.to_datetime('20150502')
print(a.day_name())
# Dataframe的 日期转换
df =pd.DataFrame({
    'sales':[100,200,300],
    'date':['20250601','20250602','20250603'],
})
df['datetime']= pd.to_datetime(df['date'])
print(df['datetime'].dt.day_name()) # series要加dt
```

```python
# csv 的日期转换
df= pd.read_excel('datest/附件1.xlsx',parse_dates=['date']) #parse将date解析为日期类型
df.info()#查看所有类型
df['datetime']= pd.to_datetime(df['date'])
```

```python
# 日期数据作为索引
df.set_index('date',inplace=True) # 将原来的date转换为索引
print(df)
print(df.loc['20250601':'20250701']) #获取6月到7月的所有数据
```

```python
# 时间的间隔
d1=pd.Timestamp('2015-05-02')
d2=pd.Timestamp('2023-05-10')
d3 =d2-d1
print(d3)
df['delta'] = df['date']-df['date'][0] # 生成一列显示与第一行的时间间隔
df.set_index('delta',inplace=True)
print(df)
print(df.loc['10days':'20days'])

days = pd.date_range('20250601','20250701',freq='D')
days = pd.date_range('20250601',periods=10,freq='YE') #periods表示要多少个数据
print(days)
```

```python
# 分组聚合
# df.groupby('分组的字段')['聚合的字段'].聚合函数（）
df=pd.read_excel('datset/附件4.xlsx')
df['体重 (kg)'] = df['体重 (kg)'].astype('int64') # 去掉小数
df.dropna(subset=['体型']) #去缺失值
# 计算男 女的平均身高
 # 分组
df.groupby('性别').groups #查看分组
df.groupby('性别').get_group('男') #查看具体的某个分组数据
# 聚合
df2 = df.groupby('性别')[['体重 (kg)']].mean() #根据男女分类 计算平均体重
df2['体重 (kg)'] = df2['体重 (kg)'].round(1) #保留一位小数
df2 = df2.reset_index()
df2 = df.groupby('性别').get_group('男')
df2.sort_values('体重 (kg)',ascending=False)
```

```python
# # 计算不同性别不同体型的平均体重
df.groupby(['性别','体型'])[['体重 (kg)']].mean()
df[['体重 (kg)']].round(1)
```
