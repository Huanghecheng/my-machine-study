```python
# Series的创建
import pandas as pd
s = pd.Series([10,1,2,3,4,5])
# print(s)
# 自定义索引
s = pd.Series([10,1,2,3,4,5],index=['a','b','c','d','e','f'])
# print(s)
# 定义name
s = pd.Series([10,1,2,3,4,5],index=['a','b','c','d','e','f'],name='月份')
print(s)
```

```python
# 通过字典的方式创建
s = pd.Series({'a':1,'b':2,'c':3,'d':4})
print(s)
s1 = pd.Series(s,index=['a','b'])
print(s1)
```

```python
# Seriers的属性
'''属性
说明
index Series的索引对象
values Series的值
dtype或dtypes Series的元素类型
shape Series的形状
ndim Series的维度
size Series的元素个数
name Series的名称
loc[] 显式索引，按标签索引或切片
iloc[] 隐式索引，按位置索引或切片
at[] 使用标签访问单个元素
iat[] 使用位置访问单个元素
'''
# print(s.index) # 索引值
# print(s.values)
# print(s.shape,s.ndim,s.size)
s.name = 'test'
print(s.dtype,s.name)
print(s.loc['a':'c']) #显示索引
print(s.iloc[0:1]) #隐式索引
print(s.at['a']) #不能跟上面一样支持切片
print(s.iat[0])
```

```python
# 访问数据
print(s)
print(s['a'])
print(s[s<3])
print(s.head()) #显示前五行
print(s.tail()) #显示末尾五行
```

```python
# Series的常见函数
import numpy as np
s = pd.Series([10,1,2,3,4,5,np.nan,None],index=['a','b','c','d','e','f','g','h'],name='data')
print(s)
```

```python
s.head(3) #默认取前5行的数据
s.tail() #默认取后五行的数据
```

```python
# 查看所有的描述性信息
s.describe()
```

```python
# 获取元素的个数（忽略缺失值）
print(s.count())
```

```python
# 获取索引
print(s.keys()) #方法方式（带小括号的就是方法）
print(s.index) # 属性方式
```

```python
# 检查Series是否有缺失值
# print(s.isna())
s.isna()
```

```python
s.isin([4,5]) #检查每个元素是否在参数集合中
```

```python
print(s.mean()) #平均数
print(s.sum()) #总和
print(s.std()) #标准差
print(s.var()) #方差
print(s.min()) #最小值
print(s.max()) #最大值
print(s.median()) #中位数
print(s.quantile(0.25)) #1/4的数
print(s.mode()) #众数
print(s.drop_duplicates()) #去重返回表格
print(s.unique())  #去重返回列表
print(s.nunique()) #去重后的元素个数
print(s.value_counts()) #每个元素的计数
```

```python
# 排序 值索引
s.sort_values() #值排序
s.sort_index() #索引排序
```
