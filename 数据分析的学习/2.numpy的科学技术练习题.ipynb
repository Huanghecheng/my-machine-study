```python
import numpy as np
from numpy.testing.print_coercion_tables import print_new_cast_table
```

```python
# 第一题
arr = np.array([28,30,29,31,32,30,29])
print('%.3f'%np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.greater(arr,29))
print(len(arr[arr>=30]))
print(np.where(arr>30,arr,0))
print(len(arr[np.where(arr>=30)]))
```

```python
# 第二题
math_score = np.array([85,90,78,92,88])
print(np.mean(math_score)) #平均值
print(np.sort(math_score))
print(np.median(math_score)) # 中位数
print(np.percentile(math_score,50))
print(np.var(math_score)) # 方差
print(np.std(math_score)) #标准差
print(arr/10) #除以10
```

```python
# 第三题
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print(A+B)
print(A*B)
print(A@B)
```

```python
# 第四题
np.random.seed(0)
arr = np.random.randint(0,10,(3,4))
print(arr)

print(np.max(arr,axis=0)) #求每一列的最大值
print(np.min(arr,axis=1)) #求每一行的最大值
print(np.where(arr%2==0,arr,-1))
arr[arr%2==1]=-1  #索引
print(arr)
```

```python
# 第五题
arr = np.arange(1,13)
print(arr)
arr = np.reshape(arr,(3,4))
print(np.sum(arr,axis=0)) #每列
print(np.sum(arr,axis=1)) #每行
print(np.mean(arr,axis=0))
print(np.mean(arr,axis=1))
arr = np.reshape(arr,(12))
print(arr)
```

```python
# 第六题
np.random.seed(0)
arr =np.random.randint(0,20,(5,5))
print(arr)
print(arr[arr>10])
print(np.where(arr>10,0,arr))
arr = arr[arr>10]
print(arr)
```

```python
# 第七题
money = np.array([120,135,110,125,130,140])
print(np.sum(money))
print(np.mean(money))
print(np.var(money))
print(np.max(money),np.argmax(money))
print(np.min(money),np.argmin(money))
```

```python
# 第八题
A = np.array([1,2,3])
B = np.array([4,5,6])
print(np.concatenate([A,B],axis=0)) #列
c = np.concatenate([A,B],axis=0)
print(np.reshape(c,(2,3)))
```

```python
# 第九题
arr =np.array([2,1,2,3,1,4,3])
print(arr)
c,counts= np.unique(arr,return_counts=True)
print(c)
print(counts)
print(np.sort(c))
```

```python
# 第二十题
m1 = np.array([20,25,22,30,28])
m2 = np.array([15,18,16,22,20])
m3 = m1 - m2
print(m3)
print(np.mean(m3)) #平均值
print('%.1f'%np.std(m3)) #标准差
print(np.max(m3),np.argmax(m3))
```
