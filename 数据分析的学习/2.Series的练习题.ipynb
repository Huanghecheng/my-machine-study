```python
import numpy as np
import pandas as pd
#第一题
  #生成随机成绩
np.random.seed(42)
scores = pd.Series(np.random.randint(50,101,10),index=['学生'+str(i)for i in range(1,11)])
print(scores)
# print(scores.mean())
# print(scores.max())
# print(scores.min())
len(scores[scores>scores.mean()]) #高于平均分的人数
```

```python
# 第二题
temperatures = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
print(len(temperatures[temperatures>30]))
print(temperatures.mean())
temperatures.sort_values(ascending=False)
t3= temperatures.diff().abs() #计算后面减去前面变化值 abs计算绝对值
t4 = t3.sort_values(ascending=False).keys()
print(t4[:2])
```

```python
# 第三题
prices = pd.Series([102.3,103.5,104.8,106.2,107.0,106.5,108.1,109.3,110.2],index=pd.date_range('2023-01-01',periods=9))
# 日期序列
data = pd.date_range('2023-01-01',periods=60)
'''
计算收益率（当前价格/前日收盘价-1
计算波动率（收益率标准差）
'''
# 计算收益率
P1 = prices.pct_change()
#收益率最高
print(P1.idxmax())
print(P1.idxmin())
print(P1.std())
```

```python
# 第四题
time = pd.date_range('2022-01-01',periods=12,freq='MS') #freq=ms是按月分布
sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=time)
print(sales)
'''
j计算季度平均销量
计算销量最高的月份
计算月环比增长率
找出连续增长超过两个月的月份
'''
sales.resample('QS').mean() #三个月季度销量
sales.idxmax()
a= sales.pct_change() #月环比增长率
b = a>0
b[b.rolling(3).sum()==3].keys().tolist() #算法 滑动窗口
```

```python
# 第五题
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0,100,24),index=pd.date_range('2025-01-01',periods=24,freq='h'))
print(hourly_sales)
# 按天重采样计算每日总营销额
zg = hourly_sales.resample('D').sum()
# 计算每天营业时间(8;00-22:00)和非营业时间的销售额比例
y1 = hourly_sales.between_time('8:00:00','22:00:00') #筛选时间
y2 = hourly_sales.drop(y1.index)
print(y1/zg.item())
print((zg-y1)/zg.item())
print(y2.sum()/zg.item())
#找出销售额最高的三个小时
hourly_sales.sort_values(ascending=False).head(3)
```
