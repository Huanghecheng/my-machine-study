```python
# dataframe的创建方式
import pandas as pd
import numpy as np
# 通过series来创建
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
df = pd.DataFrame({'第一列':s1,'第二列':s2})
# 通过字典创建
df = pd.DataFrame({
    'id':[1,2,3,4,5],
    'name':['tom','jack','alice','bob','allen'],
    'age':[15,17,20,26,30],
    'score':[60,70,80,90,10],
},index=[1,2,3,4,5],columns=['id','name','score','age'] #修改顺序
)
df
```

```python
# DataFrame的属性
print('行索引')
print(df.index)
print('列索引')
print(df.columns)
print('值')
print(df.values)
```

```python
print('维度',df.ndim)
print('数据类型',df.dtypes)
print('形状',df.shape)
print('元素个数',df.size)
# 行列转置
print(df.T.index)
```

```python
# 获取元素 loc iloc at iat
# 某行
print(df.loc[4])
print(df.iloc[3])
```

```python
# 某列
print(df.loc[:,'name'])
print(df.iloc[:,0])
```

```python
# 获取单个元素
print(df.at[3,'score']) # 行 列
print(df.iat[2,1])
print(df.loc[3,'score'])
print(df.iloc[2,1])
```

```python
# 获取单列数据
print(df['name'])
print(type(df['name']))
print(df.name)
print(type(df.name))
print(df[['name']])
print(type(df[['name']]))  #dataframe的类型
print(df[['name','score']]) # 获取多列数据
```

```python
# 查看部分数据
print(df.head(2)) # 获取前两行的数据
print(df.tail(2))
```

```python
# 使用布尔索引来筛选数据
df[df.score>70]
df[(df.score>70)&(df.age<30)]
```

```python
# 随机抽样
df.sample()
```

```python
# 通过字典创建
df = pd.DataFrame({
    'id':[1,2,3,4,5,6],
    'name':['tom','jack','alice','bob','allen','h2'],
    'age':[15,17,20,26,30,30],
    'score':[60,70,80,90,10,10],
},index=[1,2,3,4,5,6],columns=['id','name','score','age'] #修改顺序
)
# dataframe的方法
print(df.head()) # 查看前五行
print(df.tail()) # 查看后五行
print(df.isin(['jack',20,80])) # 查看元素是否在df里面
print(df.score.isna()) #查看是否为缺失值
print(df['score'].sum()) #求和
print(df.score.max())  #.是 的 意思
print(df.score.min())
print(df.score.mean()) #平均数
print(df.score.std())
print(df.score.var()) #方差
print(df.score.median()) # 中位数
print(df.score.mode()) # 众数
print(df.score.quantile()) # 分位数
print(df.describe()) # 拿到数据可以用这样来描述
print(df.count()) # 每一列非缺失值的个数
```

```python
print(df.value_counts()) # 重复值出现的次数
print(df.drop_duplicates()) # 去重复值
print(df.duplicated(subset=['score']))  #检查是否重复 subset可以指定检查
df.sample(2) # 随机采2个样
print(df.replace(15,20)) # 将15换成30 替换
df.cumsum() # 累加
df.cummax() #累积的最大
df.cummin(axis=0) # 0按列求 1按行
```

```python
print(df.sort_values(by='score',ascending=False)) # False是从高到低排序
print(df.sort_values(by=['score','age']))# 从低到高排序
```

```python
df.nlargest(1,columns=['score']) # 取n个最大的数据 从大到小
df.nsmallest(1,columns=['score'])# 取n个最小的数据  从小到大
```
