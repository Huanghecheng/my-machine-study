```python
# 缺失值的处理
import pandas as pd
import numpy as np
s = pd.Series([1,2,3,4,np.nan,pd.NA,None])
df = pd.DataFrame({
    '人类':['我','你','他'],
    '成绩':[90,np.nan,70]
})

# 查看是否为缺失值
print(s.isna())
print(s.isnull())
print(df.isnull())
print(df.isna().sum(axis=1)) # 查看缺失值的个数 0按列数
print('-'*10)
# 删除缺失值
print(df.dropna()) # 删除一整条数据
print(df.dropna(how='all')) # 如果所有的值都是缺失值 才删除
print(df.dropna(thresh=1))  # 如果至少有n个值不是缺失值 就保留
print(df.dropna(axis=0))  # 删除一整列的记录
print(df.dropna(subset=['人类'])) #如果某列有缺失值 就删除这一行
print('-'*10)

# 填充缺失值
df = pd.read_excel('datset/附件2.xlsx',engine='openpyxl')
df.isna().sum()
print(df.fillna({'销售日期':20,'销售单价':10})) # 填充某列缺失值
print(df.fillna(df[['销售日期','销售单价(元/千克)']].mean(axis=0)).tail()) # 填充平均值
print(df.ffill()) # 用前面的值填充
print(df.bfill()) # 使用后面的相邻值填充
```

```python
# 数据的变形
import pandas as pd
df = pd.DataFrame({
    'ID':[2,3],
    'name':['alice','bob'],
    'Math':[90,85],
    'English':[88,92],
    'Science':[95,89]
})
df.T # 行列转置
print(df)
'''
1 alice math 90
1 alice English 88
1 alice Science 95
'''
# 宽表转变为长表
df2 = pd.melt(df,id_vars=['ID','name'],var_name='科目',value_name='分数')  # 将一个宽表变成一个长表
df2.sort_values('name')
# 长表转为宽表
pd.pivot(df2,index=['ID','name'],columns='科目',values='分数')
```

```python
# 数据分箱 pd.cut(x,bins,labels)
import pandas as pd
df = pd.read_excel('datset/附件4.xlsx',engine='openpyxl')
df1 = df[['姓名','身高 (cm)']]
print(df1)
pd.cut(df1['身高 (cm)'],bins=2) # bins = n 表示分成了n个区间，起始值，结束值是所有数据的最小值和最大值
# (146.75, 173.5]和(119.946, 146.75]
pd.cut(df1['身高 (cm)'],bins=2).value_counts() # 各个区域里面有多少
pd.cut(df1['身高 (cm)'],bins=[0,160,170]).value_counts() #bins=list,分成n段区间
df1['身高'] = pd.cut(df1['身高 (cm)'],bins=[0,160,170],labels=['低','高']) # 将数据转为选择
pd.qcut(df1['身高 (cm)'],3).value_counts() # 将人数平均分
```

```python
import pandas as pd
df = pd.read_excel('datset/附件4.xlsx',engine='openpyxl')
df = df[['姓名','性别','身高 (cm)','体重 (kg)']]
df['BMI'] =(df['体重 (kg)']) / (df['身高 (cm)'] /100)**2
df
df['mbi是否正常'] = pd.cut(df['BMI'],bins=[0,18.5,23.9],labels=['瘦','正常'])
df['性别'].astype('category' ) # 将它数据转换成选择
df['性别'] = df['性别'].astype('category' )
df['性别'].value_counts() #查看男女个数
# 字符串--> 类别-->统计
# 数值-->分箱-->统计
```
