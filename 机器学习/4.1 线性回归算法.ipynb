## 1.生成用于回归分析的有1个特征的数据集

```python
import numpy as np
import matplotlib.pyplot as plt
# 导入回归数据生成器与线性回归模型
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
# 导入均方误差评估模块
from sklearn.metrics import mean_squared_error
# 生成用于回归分析的数据集
X,y = make_regression(n_samples=50,n_features=1,n_informative=1,random_state=3,noise=50)
```

## 2.使用线性回归对数据进行拟合

```python
# 使用线性回归模型对数据进行拟合
import sklearn
sklearn.set_config(display="text")

lr = LinearRegression()
lr.fit(X,y)
```

## 3.画出线性回归模型的图形

```python
z = np.linspace(-3,3,100).reshape(-1,1)
plt.scatter(X,y,color="blue",s=60)
plt.plot(z,lr.predict(z),color="purple")

plt.rcParams['font.sans-serif']=['SimHei']
plt.rcParams['axes.unicode_minus']=False
plt.title('线性回归模型')
```

## 4.输出线性回归模型的系数，截距和方程

```python
print('斜率：{:.2f}'.format(lr.coef_[0]))
print('截距：{:.2f}'.format(lr.intercept_))
print('直线的方程：{:.2f}'.format(lr.coef_[0],'x','+{:.2f}'.format(lr.intercept_)))
```

## 5.生成用于回归分析的有2个特征的数据集

```python
# 导入数据集拆分工具
from sklearn.model_selection import train_test_split
# 生成用于回归分析的有2个特征的数据集
X,y = make_regression(n_samples=100,n_features=2,n_informative=2,random_state=38)
```

## 6.使用线性回归模型对数据拟合，并输出线性回归模型的系数，截距

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=8)
lr2 = LinearRegression().fit(X_train,y_train)
y_predict = lr2.predict(X_test)
print('lr.coef_:{}'.format(lr2.coef_[:]))
print('lr.intercept_:{}'.format(lr.intercept_))
```

## 7.线性回归模型的性能评估

```python
mse = mean_squared_error(y_test,y_predict)
print('均方误差为：\n',mse)
```
